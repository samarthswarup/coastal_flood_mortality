```python
import xarray as xr
import zipfile
```

```python
with zipfile.ZipFile('./slr_allscens_byseg.nc.zip', 'r') as zip_ref:
    zip_ref.extractall()
slr = xr.open_dataset('./slr_allscens_byseg.nc')
slr
```

```python
### Select any sea level rise projection and year here from above scenario ######
### Here I have chosen K14_rcp85 for year 2050 #########
selected_value = slr['lsl_msl05'].sel(quantile=0.83, scenario='K14_rcp85', year=2050)
```

```python
selected_df = selected_value.to_dataframe()
```

```python
selected_df.to_csv('./k14_rcp85_2050.csv')
```

```python
selected_value = slr['lsl_msl05'].sel(quantile=0.83, scenario='K14_rcp85', year=2100)
selected_df = selected_value.to_dataframe()
selected_df.to_csv('./k14_rcp85_2100.csv')
```
