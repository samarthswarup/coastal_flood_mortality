```python
import xarray as xr
import zipfile
```

```python
with zipfile.ZipFile('./slr_allscens_byseg.nc.zip', 'r') as zip_ref:
    zip_ref.extractall()
slr = xr.open_dataset('./slr_allscens_byseg.nc')
slr
```

```python
scenarios = ['AR6_low_ssp126', 'AR6_low_ssp585', 'AR6_med_ssp119', 'AR6_med_ssp126', 'AR6_med_ssp245', 'AR6_med_ssp370', 'AR6_med_ssp585']
quantiles = [0.17, 0.5, 0.83]
years = [2050, 2100]
for s in scenarios:
    for q in quantiles:
        for y in years:
            selected_value = slr['lsl_msl05'].sel(quantile=q, scenario=s, year=y)
            selected_df = selected_value.to_dataframe()
            q100 = q*100
            filename = f'./{s}_{y}_{q100}.csv'
            selected_df.to_csv(filename)
```

```python
selected_value = slr['lsl_ncc_msl05'].sel(scenario=scenarios, quantile=quantiles, year=years) #Select only 2050 and 2100 for year
selected_df = selected_value.to_dataframe()
selected_df.to_csv('NCC_AR6.csv')
```
